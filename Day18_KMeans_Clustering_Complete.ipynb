## 1. Import Libraries

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

## 2. Load Dataset

We'll use the Iris dataset as a real-world dataset. The labels are ignored because K-Means is an unsupervised algorithm.


```
iris=load_iris()
X=pd.DataFrame(iris.data,columns=iris.feature_names)

print(X.head())
print(X.shape)
```

## 3. Feature Scaling

```
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

## 4. Elbow Method

```
inertia=[]
K=range(1,11)

for k in K:
    model=KMeans(n_clusters=k,random_state=42,n_init=10)
    model.fit(X_scaled)
    inertia.append(model.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K,inertia,marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid(True)
plt.show()
```

## 5. Train K-Means

We'll choose **K = 3** because it matches the elbow and Iris naturally has three groups.


```
kmeans=KMeans(
    n_clusters=3,
    init='k-means++',
    n_init=10,
    max_iter=300,
    random_state=42
)

labels=kmeans.fit_predict(X_scaled)
centroids=kmeans.cluster_centers_

print("Centroids")
print(centroids)
print("\nIterations:",kmeans.n_iter_)
print("Inertia:",kmeans.inertia_)
```

## 6. Visualize Clusters

```
plt.figure(figsize=(8,6))

plt.scatter(X_scaled[:,0],X_scaled[:,1],c=labels,s=60)
plt.scatter(centroids[:,0],centroids[:,1],
            marker='X',s=250)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("K-Means Clusters")
plt.show()
```

## 7. Silhouette Score

```
score=silhouette_score(X_scaled,labels)
print("Silhouette Score:",round(score,3))
```

## 8. Cluster Summary

```
result=X.copy()
result["Cluster"]=labels
display(result.groupby("Cluster").mean())
```

# Conclusion

- K-Means groups similar observations.
- Scaling is essential.
- Elbow Method helps estimate K.
- Silhouette Score evaluates cluster quality.
- K-Means++ provides better initialization.
- K-Means works best for compact, spherical clusters.
