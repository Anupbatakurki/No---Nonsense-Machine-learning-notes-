## Story & Intuition

Imagine finding the only red apple in a basket of green apples. Random questions isolate the unusual item quickly. Isolation Forest uses the same idea.

## Algorithm Steps

1. Build many random trees.
2. Randomly choose features and split values.
3. Measure path length.
4. Short path = anomaly.

```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from sklearn.metrics import confusion_matrix, classification_report

np.random.seed(42)
```

## Create Dataset

```
X_normal,_=make_blobs(n_samples=500,centers=3,cluster_std=0.7,random_state=42)
X_out=np.random.uniform(-8,8,size=(30,2))
X=np.vstack([X_normal,X_out])
y_true=np.array([0]*500+[1]*30)
plt.figure(figsize=(7,6))
plt.scatter(X[:,0],X[:,1],alpha=.7)
plt.title("Original Dataset")
plt.show()
```

## Scale Features

```
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

## Train Isolation Forest

```
model=IsolationForest(
    n_estimators=200,
    contamination=30/530,
    random_state=42,
    n_jobs=-1)
model.fit(X_scaled)
```

## Predict

```
pred=model.predict(X_scaled)
pred_binary=np.where(pred==-1,1,0)
scores=model.decision_function(X_scaled)

print("Anomalies detected:",pred_binary.sum())
```

## Prediction Visualization

```
normal=pred==1
anom=pred==-1

plt.figure(figsize=(8,6))
plt.scatter(X[normal,0],X[normal,1],label="Normal",alpha=.7)
plt.scatter(X[anom,0],X[anom,1],marker="x",s=120,label="Anomaly")
plt.legend()
plt.title("Isolation Forest Prediction")
plt.show()
```

## Decision Function

```
plt.figure(figsize=(8,6))
im=plt.scatter(X[:,0],X[:,1],c=scores,cmap="RdYlBu")
plt.colorbar(im,label="Decision Function")
plt.title("Anomaly Score Heatmap")
plt.show()
```

## Evaluate (Synthetic Labels)

```
print(confusion_matrix(y_true,pred_binary))
print(classification_report(y_true,pred_binary))
```

## Approximate Feature Importance

```
importance=np.zeros(X.shape[1])
for tree in model.estimators_:
    for f in tree.tree_.feature:
        if f>=0:
            importance[f]+=1
importance/=importance.sum()

plt.figure(figsize=(5,4))
plt.bar(["Feature 1","Feature 2"],importance)
plt.title("Approximate Feature Importance")
plt.show()
print(importance)
```
