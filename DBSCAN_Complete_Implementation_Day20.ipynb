# DBSCAN (Day 20) - Complete Implementation

A complete notebook explaining DBSCAN with practical examples.

## Imports

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons, make_circles, load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN, KMeans
from sklearn.metrics import silhouette_score
from sklearn.neighbors import NearestNeighbors
```

## 1. DBSCAN vs K-Means

```
X,_=make_moons(n_samples=300,noise=0.08,random_state=42)
X=StandardScaler().fit_transform(X)
km=KMeans(n_clusters=2,n_init=10,random_state=42)
km_labels=km.fit_predict(X)
db=DBSCAN(eps=0.3,min_samples=5)
db_labels=db.fit_predict(X)
fig,ax=plt.subplots(1,2,figsize=(10,4))
ax[0].scatter(X[:,0],X[:,1],c=km_labels)
ax[0].set_title("K-Means")
ax[1].scatter(X[:,0],X[:,1],c=db_labels)
ax[1].set_title("DBSCAN")
plt.show()
print("Clusters:",len(set(db_labels))-(1 if -1 in db_labels else 0))
print("Noise:",list(db_labels).count(-1))
```

## 2. Effect of eps

```
X,_=make_circles(n_samples=400,factor=0.5,noise=0.05,random_state=42)
X=StandardScaler().fit_transform(X)
eps_values=[0.1,0.2,0.3,0.5]
fig,axs=plt.subplots(2,2,figsize=(10,8))
for ax,eps in zip(axs.ravel(),eps_values):
    labels=DBSCAN(eps=eps,min_samples=5).fit_predict(X)
    ax.scatter(X[:,0],X[:,1],c=labels)
    ax.set_title(f'eps={eps}')
plt.tight_layout(); plt.show()
```

## 3. K-Distance Graph

```
nbrs=NearestNeighbors(n_neighbors=5).fit(X)
dist,_=nbrs.kneighbors(X)
plt.figure(figsize=(7,4))
plt.plot(np.sort(dist[:,4]))
plt.grid(True)
plt.title("K-Distance Graph")
plt.show()
```

## 4. Real Dataset

```
wine=load_wine()
X=StandardScaler().fit_transform(wine.data)
labels=DBSCAN(eps=2.2,min_samples=5).fit_predict(X)
clusters=len(set(labels))-(1 if -1 in labels else 0)
noise=list(labels).count(-1)
print("Clusters:",clusters)
print("Noise:",noise)
mask=labels!=-1
if clusters>1 and mask.sum()>2:
    print("Silhouette:",silhouette_score(X[mask],labels[mask]))
```

## 5. Noise Detection

```
np.random.seed(42)
c1=np.random.randn(120,2)*0.4+[2,2]
c2=np.random.randn(120,2)*0.4+[-2,-2]
n=np.random.uniform(-5,5,(50,2))
X=np.vstack([c1,c2,n])
X=StandardScaler().fit_transform(X)
labels=DBSCAN(eps=0.3,min_samples=5).fit_predict(X)
mask=labels==-1
plt.figure(figsize=(7,5))
plt.scatter(X[~mask,0],X[~mask,1],c=labels[~mask],cmap='tab10')
plt.scatter(X[mask,0],X[mask,1],marker='x')
plt.title("Noise Detection")
plt.show()
```

## Conclusion
- DBSCAN finds dense regions.
- Detects outliers automatically.
- Does not require specifying K.
- eps and min_samples control clustering quality.
