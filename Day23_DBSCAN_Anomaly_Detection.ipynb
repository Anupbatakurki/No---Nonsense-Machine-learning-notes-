# Day 23 - DBSCAN for Anomaly Detection
## Beginner-Friendly Notebook

### Import Libraries

```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.ensemble import IsolationForest
```

### Generate Dataset

```
np.random.seed(42)
X,_=make_blobs(n_samples=300,centers=3,cluster_std=0.6,random_state=42)
noise=np.random.uniform(-6,6,(30,2))
X=np.vstack([X,noise])
plt.scatter(X[:,0],X[:,1]); plt.title("Original Data"); plt.show()
```

### Scale Data

```
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

### Train DBSCAN

```
db=DBSCAN(eps=0.5,min_samples=5)
labels=db.fit_predict(X_scaled)
print("Clusters:",len(set(labels))-(1 if -1 in labels else 0))
print("Anomalies:",(labels==-1).sum())
```

### Visualize

```
mask=labels!=-1
plt.figure(figsize=(6,5))
plt.scatter(X[mask,0],X[mask,1],c=labels[mask],cmap="tab10")
plt.scatter(X[~mask,0],X[~mask,1],c="red",marker="x",s=80)
plt.title("DBSCAN Anomaly Detection")
plt.show()
```

### Effect of eps

```
fig,axs=plt.subplots(2,2,figsize=(10,8))
for ax,eps in zip(axs.ravel(),[0.3,0.5,0.7,1.0]):
    lab=DBSCAN(eps=eps,min_samples=5).fit_predict(X_scaled)
    m=lab!=-1
    ax.scatter(X[m,0],X[m,1],c=lab[m],cmap="tab10",s=15)
    ax.scatter(X[~m,0],X[~m,1],c="red",marker="x")
    ax.set_title(f"eps={eps}")
plt.tight_layout(); plt.show()
```

### DBSCAN vs Isolation Forest

```
iso=IsolationForest(contamination=0.1,random_state=42)
pred=iso.fit_predict(X_scaled)
fig,ax=plt.subplots(1,2,figsize=(12,5))
m=labels!=-1
ax[0].scatter(X[m,0],X[m,1]); ax[0].scatter(X[~m,0],X[~m,1],c='r',marker='x'); ax[0].set_title("DBSCAN")
m2=pred==1
ax[1].scatter(X[m2,0],X[m2,1]); ax[1].scatter(X[~m2,0],X[~m2,1],c='r',marker='x'); ax[1].set_title("Isolation Forest")
plt.show()
```

## Summary

- DBSCAN groups dense regions into clusters.
- Points labeled **-1** are treated as anomalies.
- Scaling is important because DBSCAN relies on distance.
- Tune **eps** and **min_samples** carefully.
